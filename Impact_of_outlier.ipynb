```python
import numpy as np
A = np.array([[0,1],[1,4],[2,3],[5,6]])
b= np.array([[2],
             [9],
             [8],
             [39]])
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression().fit(A, b)
y_pred = reg.predict(A)

from matplotlib import pyplot as plt
plt.figure()
plt.scatter(A[:,0],b)
plt.plot(A[:,0],y_pred)
plt.legend(['Original','Predicted'])
plt.xlabel('test_feature 1')
plt.ylabel('Output')
plt.title('Comparison between 1st feature and label')
```
